```python
"""
Shared color palette for consistent solver colors across all plots
"""

# Define a consistent color mapping for each solver
SOLVER_COLORS = {
    'euler': '#5F8D8E',                    # Muted Teal
    'shark': '#C07F66',                    # Terra Cotta
    'reversibleheun': '#6B8E6F',           # Sage
    'stratonovichmilstein': '#D4B46A',     # Mustard
    'itomilstein': '#8C7B99',              # Dusty Purple
    'ees25': '#5D6D7E',                    # Slate
    'eulerheun': '#4A5D4F',                # Dark Forest
    'heun': '#B8868A',                     # Dusty Rose
    'midpoint': '#A67B5B',                 # Clay
}

# Fallback colors for any solvers not in the mapping
FALLBACK_COLORS = [
    '#7A9D8F', '#D4A574', '#9B7B9E', '#6B8E8F',
    '#C89F77', '#8E7B99', '#5F8E8D', '#B88A86'
]

def get_solver_color(solver_name, index=0):
    """
    Get consistent color for a solver by name

    Args:
        solver_name: Name of the solver
        index: Fallback index if solver not in mapping

    Returns:
        str: Hex color code
    """
    # Normalize solver name (lowercase, strip whitespace)
    normalized_name = solver_name.lower().strip()

    # Try exact match first
    if normalized_name in SOLVER_COLORS:
        return SOLVER_COLORS[normalized_name]

    # Try partial match (in case of variations)
    for key in SOLVER_COLORS:
        if key in normalized_name or normalized_name in key:
            return SOLVER_COLORS[key]

    # Fallback to index-based color
    return FALLBACK_COLORS[index % len(FALLBACK_COLORS)]

def get_color_palette(solver_names):
    """
    Get a list of colors for a list of solver names

    Args:
        solver_names: List of solver names

    Returns:
        list: List of hex color codes
    """
    return [get_solver_color(name, idx) for idx, name in enumerate(solver_names)]
```
