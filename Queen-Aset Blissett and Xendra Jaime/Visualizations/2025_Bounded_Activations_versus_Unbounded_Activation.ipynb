#Summary of Code

This code creates an informative video using the ManimCE library. This is done to explain three different activation functions, namely Sigmoid, Tanh, and ReLU, and specifically focus on their 'bounded interval' or 'unbounded interval' nature. The video then discusses the implications on these chracteristics and their associated properties for the Universal Approximation Theorem. This is a new visualisation done for the 2025 Proceedings.

```python
from manim import *
import numpy as np
```

## Setting up colors

```python
BLUEISHWHITE = ManimColor.from_hex("#F5F5FA")
NEWBLUE = ManimColor.from_hex("#2d78e0")
NEWDARKBLUE = ManimColor.from_hex("#050533")
NEWMEDIUMBLUE = ManimColor.from_hex("#404AA8")
```

## The full code

```python
class ActivationFunctionsExplainer(Scene):
    def construct(self):
        #Set the background colour.
        self.camera.background_color = NEWDARKBLUE

        #Define the axes and set the colour of the axes.
        axes = Axes(
            x_range=[-5, 6, 1],
            y_range=[-3, 3, 1],
            axis_config={"color": BLUEISHWHITE}
        )

        # SCENE 1: Initial Questions
        self.scene_1_questions()
        self.wait(1)
        self.clear()

        # SCENE 2: Function Presentation
        self.scene_2_presentation()
        self.wait(1)
        self.clear()

        # SCENE 3: Sigmoid Analysis
        self.scene_3_sigmoid()
        self.wait(1)
        self.clear()

        # SCENE 4: Tanh Analysis
        self.scene_4_tanh()
        self.wait(1)
        self.clear()

        # SCENE 5: ReLU Analysis
        self.scene_5_relu()
        self.wait(1)
        self.clear()

        # SCENE 6: Quick Comparison
        self.scene_6_comparison()
        self.wait(1)
        self.clear()

        # SCENE 7: UAT Importance
        self.scene_7_uat()
        self.wait(1)
        self.clear()

        # SCENE 8: Conclusion
        self.scene_8_conclusion()
        self.wait(2)

    def scene_1_questions(self):
        """Scene 1: Initial questions"""
        question1 = Text("What does it mean that \nan activation function is bounded?",
                        font_size=40).move_to(ORIGIN)

        self.play(FadeIn(question1))
        self.wait(2)
        self.play(FadeOut(question1))
        self.wait(0.5)

        question2 = Text("And why does it matter for the \nUniversal Approximation Theorem?",
                        font_size=36).move_to(ORIGIN)

        self.play(FadeIn(question2))
        self.wait(2)
        self.play(FadeOut(question2))

    def scene_2_presentation(self):
        """Scene 2: Function presentation"""
        sigmoid_title = Text("Sigmoid", font_size=36, color=BLUE)
        tanh_title = Text("Tanh", font_size=36, color=GREEN)
        relu_title = Text("ReLU", font_size=36, color=RED)

        # Sigmoid appears and moves left
        sigmoid_title.move_to(ORIGIN)
        self.play(Write(sigmoid_title))
        self.wait(0.5)
        self.play(sigmoid_title.animate.shift(LEFT * 4))

        # Tanh appears in center
        tanh_title.move_to(ORIGIN)
        self.play(Write(tanh_title))
        self.wait(0.5)

        # ReLU appears and moves right
        relu_title.move_to(ORIGIN)
        self.play(Write(relu_title))
        self.wait(0.5)
        self.play(relu_title.animate.shift(RIGHT * 4))

        # Add equations
        sigmoid_eq = MathTex(r"\sigma(x) = \frac{1}{1+e^{-x}}",
                            font_size=24, color=BLUE).next_to(sigmoid_title, DOWN)
        tanh_eq = MathTex(r"\tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}",
                         font_size=24, color=GREEN).next_to(tanh_title, DOWN)
        relu_eq = MathTex(r"\text{ReLU}(x) = \max(0, x)",
                         font_size=24, color=RED).next_to(relu_title, DOWN)

        self.play(
            FadeIn(sigmoid_eq),
            FadeIn(tanh_eq),
            FadeIn(relu_eq)
        )
        self.wait(2)

    def scene_3_sigmoid(self):
        """Scene 3: Sigmoid analysis with zooming"""
        axes = Axes(
            x_range=[-6, 6, 1],
            y_range=[-0.5, 1.5, 0.5],
            x_length=10,
            y_length=6,
            axis_config={"color": BLUEISHWHITE}
        )

        # Sigmoid function
        sigmoid = axes.plot(lambda x: 1 / (1 + np.exp(-x)),
                           x_range=[-6, 6], color=BLUE)

        title = Text("Sigmoid Function", font_size=32, color=BLUE).to_edge(UP)

        self.play(Create(axes), Write(title))
        self.play(Create(sigmoid))
        self.wait(1)

        # Add dashed bounds
        upper_bound = DashedLine(
            axes.c2p(-6, 1), axes.c2p(6, 1), color=YELLOW
        )
        lower_bound = DashedLine(
            axes.c2p(-6, 0), axes.c2p(6, 0), color=YELLOW
        )

        self.play(Create(upper_bound), Create(lower_bound))
        self.wait(1)

        # Zoom to right tail
        zoom_text = Text("Going to the right...",
                        font_size=24).to_edge(DOWN)
        self.play(Write(zoom_text))

        # Create moving point
        x_tracker = ValueTracker(0)
        dot = always_redraw(lambda: Dot(
            axes.c2p(x_tracker.get_value(),
                    1 / (1 + np.exp(-x_tracker.get_value()))),
            color=RED
        ))

        x_label = always_redraw(lambda: MathTex(
            f"x = {x_tracker.get_value():.1f}",
            font_size=20
        ).next_to(dot, UP))

        self.play(FadeIn(dot), FadeIn(x_label))
        self.play(x_tracker.animate.set_value(5), run_time=3, rate_func=smooth)

        upper_label = Text("Upper bound: 1", font_size=24, color=YELLOW).next_to(
            axes.c2p(3, 1), UP
        )
        self.play(Write(upper_label))
        self.wait(1)

        # Zoom to left tail
        self.play(FadeOut(zoom_text))
        zoom_text_left = Text("Going to the left...",
                             font_size=24).to_edge(DOWN)
        self.play(Write(zoom_text_left))

        self.play(x_tracker.animate.set_value(-5), run_time=3, rate_func=smooth)

        lower_label = Text("Lower bound: 0", font_size=24, color=YELLOW).next_to(
            axes.c2p(-3, 0), DOWN
        )
        self.play(Write(lower_label))
        self.wait(2)

    def scene_4_tanh(self):
        """Scene 4: Tanh analysis"""
        axes = Axes(
            x_range=[-6, 6, 1],
            y_range=[-1.5, 1.5, 0.5],
            x_length=10,
            y_length=6,
            axis_config={"color": BLUEISHWHITE}
        )

        # Tanh function
        tanh = axes.plot(lambda x: np.tanh(x),
                        x_range=[-6, 6], color=GREEN)

        title = Text("Tanh Function", font_size=32, color=GREEN).to_edge(UP)

        self.play(Create(axes), Write(title))
        self.play(Create(tanh))
        self.wait(1)

        # Add dashed bounds
        upper_bound = DashedLine(
            axes.c2p(-6, 1), axes.c2p(6, 1), color=YELLOW
        )
        lower_bound = DashedLine(
            axes.c2p(-6, -1), axes.c2p(6, -1), color=YELLOW
        )

        self.play(Create(upper_bound), Create(lower_bound))
        self.wait(1)

        # Moving point to +infinity
        x_tracker = ValueTracker(0)
        dot = always_redraw(lambda: Dot(
            axes.c2p(x_tracker.get_value(),
                    np.tanh(x_tracker.get_value())),
            color=RED
        ))

        x_label = always_redraw(lambda: MathTex(
            f"x = {x_tracker.get_value():.1f}",
            font_size=20
        ).next_to(dot, UP))

        self.play(FadeIn(dot), FadeIn(x_label))
        self.play(x_tracker.animate.set_value(5), run_time=2.5, rate_func=smooth)

        upper_label = Text("Upper bound: 1", font_size=24, color=YELLOW).next_to(
            axes.c2p(3, 1), UP
        )
        self.play(Write(upper_label))
        self.wait(1)

        # Moving point to -infinity
        self.play(x_tracker.animate.set_value(-5), run_time=2.5, rate_func=smooth)

        lower_label = Text("Lower bound: -1", font_size=24, color=YELLOW).next_to(
            axes.c2p(-3, -1), DOWN
        )
        self.play(Write(lower_label))
        self.wait(2)

    def scene_5_relu(self):
        """Scene 5: ReLU analysis with escaping ceiling"""
        axes = Axes(
            x_range=[-2, 6, 1],
            y_range=[-1, 7, 1],
            x_length=10,
            y_length=6,
            axis_config={"color": BLUEISHWHITE}
        )

        # ReLU function
        relu = axes.plot(lambda x: max(0, x),
                        x_range=[-2, 6], color=RED)

        title = Text("ReLU Function", font_size=32, color=RED).to_edge(UP)

        self.play(Create(axes), Write(title))
        self.play(Create(relu))
        self.wait(1)

        question = Text("Where is the 'ceiling'?",
                       font_size=28, color=YELLOW).to_edge(DOWN)
        self.play(Write(question))
        self.wait(1)

        # Escaping ceiling
        ceiling_y = ValueTracker(4)
        ceiling = always_redraw(lambda: DashedLine(
            axes.c2p(-2, ceiling_y.get_value()),
            axes.c2p(6, ceiling_y.get_value()),
            color=YELLOW
        ))

        self.play(Create(ceiling))

        # Moving point
        x_tracker = ValueTracker(0)
        dot = always_redraw(lambda: Dot(
            axes.c2p(x_tracker.get_value(),
                    max(0, x_tracker.get_value())),
            color=WHITE
        ))

        x_label = always_redraw(lambda: MathTex(
            f"x = {x_tracker.get_value():.1f}",
            font_size=20
        ).next_to(dot, DOWN))

        self.play(FadeIn(dot), FadeIn(x_label))

        # Animate point moving and ceiling escaping
        self.play(
            x_tracker.animate.set_value(5.5),
            ceiling_y.animate.set_value(10),
            run_time=3,
            rate_func=smooth
        )

        conclusion = Text("Oh... it never stops!",
                         font_size=32, color=RED).next_to(question, UP)
        self.play(Write(conclusion))
        self.wait(2)

    def scene_6_comparison(self):
        """Scene 6: Side-by-side comparison"""
        # Create three small axes
        axes_sigmoid = Axes(
            x_range=[-4, 4, 2],
            y_range=[-0.5, 1.5, 0.5],
            x_length=3.5,
            y_length=2.5,
            axis_config={"color": BLUEISHWHITE, "font_size": 16}
        ).shift(LEFT * 4 + UP * 0.5)

        axes_tanh = Axes(
            x_range=[-4, 4, 2],
            y_range=[-1.5, 1.5, 1],
            x_length=3.5,
            y_length=2.5,
            axis_config={"color": BLUEISHWHITE, "font_size": 16}
        ).shift(UP * 0.5)

        axes_relu = Axes(
            x_range=[-1, 4, 1],
            y_range=[-0.5, 4.5, 1],
            x_length=3.5,
            y_length=2.5,
            axis_config={"color": BLUEISHWHITE, "font_size": 16}
        ).shift(RIGHT * 4 + UP * 0.5)

        # Plot functions
        sigmoid = axes_sigmoid.plot(lambda x: 1 / (1 + np.exp(-x)),
                                   x_range=[-4, 4], color=BLUE)
        tanh = axes_tanh.plot(lambda x: np.tanh(x),
                             x_range=[-4, 4], color=GREEN)
        relu = axes_relu.plot(lambda x: max(0, x),
                             x_range=[-1, 4], color=RED)

        # Add bounds
        sigmoid_upper = DashedLine(
            axes_sigmoid.c2p(-4, 1), axes_sigmoid.c2p(4, 1), color=YELLOW
        )
        sigmoid_lower = DashedLine(
            axes_sigmoid.c2p(-4, 0), axes_sigmoid.c2p(4, 0), color=YELLOW
        )

        tanh_upper = DashedLine(
            axes_tanh.c2p(-4, 1), axes_tanh.c2p(4, 1), color=YELLOW
        )
        tanh_lower = DashedLine(
            axes_tanh.c2p(-4, -1), axes_tanh.c2p(4, -1), color=YELLOW
        )

        # Titles
        title_sigmoid = Text("Sigmoid", font_size=24, color=BLUE).next_to(
            axes_sigmoid, UP, buff=0.2
        )
        title_tanh = Text("Tanh", font_size=24, color=GREEN).next_to(
            axes_tanh, UP, buff=0.2
        )
        title_relu = Text("ReLU", font_size=24, color=RED).next_to(
            axes_relu, UP, buff=0.2
        )

        self.play(
            Create(axes_sigmoid), Create(axes_tanh), Create(axes_relu),
            Write(title_sigmoid), Write(title_tanh), Write(title_relu)
        )

        self.play(
            Create(sigmoid), Create(tanh), Create(relu)
        )

        self.play(
            Create(sigmoid_upper), Create(sigmoid_lower),
            Create(tanh_upper), Create(tanh_lower)
        )

        # Labels
        bounded_label = Text("Bounded activations",
                           font_size=28, color=YELLOW).move_to(DOWN * 2.5 + LEFT * 2.2)
        unbounded_label = Text("Unbounded activation",
                             font_size=28, color=YELLOW).move_to(DOWN * 2.5 + RIGHT * 2.4)

        vs_text = Text("vs", font_size=32).move_to(DOWN * 2.5)

        self.play(Write(bounded_label), Write(vs_text), Write(unbounded_label))
        self.wait(2)

    def scene_7_uat(self):
        """Scene 7: Universal Approximation Theorem"""
        title = Text("Importance for the UAT",
                    font_size=36, color=YELLOW).to_edge(UP)
        self.play(Write(title))
        self.wait(1)

        # Bounded functions combination
        axes1 = Axes(
            x_range=[-4, 4, 1],
            y_range=[-2, 2, 1],
            x_length=6,
            y_length=3,
            axis_config={"color": BLUEISHWHITE}
        ).shift(UP * 0.5)

        # Multiple sigmoid-like curves
        curves = VGroup(*[
            axes1.plot(lambda x, i=i: 0.8 * (1 / (1 + np.exp(-(x - i*0.5)))) - 0.4,
                      x_range=[-4, 4], color=BLUE, stroke_width=2)
            for i in range(-3, 4)
        ])

        explanation1 = Text("Bounded functions allow\ncontrolled approximations",
                          font_size=24).to_edge(DOWN)

        self.play(Create(axes1))
        self.play(Create(curves, lag_ratio=0.1))
        self.play(Write(explanation1))
        self.wait(2)

        # Target function approximation
        target = axes1.plot(lambda x: 0.5 * np.sin(2*x),
                          x_range=[-4, 4], color=RED, stroke_width=4)
        self.play(Create(target))
        self.wait(2)

        self.play(FadeOut(curves), FadeOut(target), FadeOut(axes1),
                 FadeOut(explanation1))

        # ReLU piecewise construction
        axes2 = Axes(
            x_range=[-4, 4, 1],
            y_range=[-2, 2, 1],
            x_length=6,
            y_length=3,
            axis_config={"color": BLUEISHWHITE}
        ).shift(UP * 0.5)

        # Piecewise linear segments
        points = [(-4, -1), (-2, 0.5), (0, 1), (2, 0.3), (4, -0.5)]
        piecewise = VGroup()
        for i in range(len(points) - 1):
            line = Line(
                axes2.c2p(points[i][0], points[i][1]),
                axes2.c2p(points[i+1][0], points[i+1][1]),
                color=RED, stroke_width=3
            )
            piecewise.add(line)

        explanation2 = Text("ReLU can also approximate...\nbut through a different path",
                          font_size=24).to_edge(DOWN)

        self.play(Create(axes2))
        self.play(Create(piecewise, lag_ratio=0.2))
        self.play(Write(explanation2))
        self.wait(2)

    def scene_8_conclusion(self):
        """Scene 8: Final conclusion"""
        conclusion1 = Text(
            "Bounded activations → piecewise constant approximators",
            font_size=32,
            color=BLUE
        ).move_to(UP * 1.5)

        conclusion2 = Text(
            "ReLU → unbounded\n→ piecewise linear approximations",
            font_size=32,
            color=RED
        ).move_to(DOWN * 1.5)

        self.play(Write(conclusion1))
        self.wait(1.5)
        self.play(Write(conclusion2))
        self.wait(3)
        self.play(FadeOut(conclusion1), FadeOut(conclusion2))
```
