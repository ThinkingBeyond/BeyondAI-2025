```python
import yfinance as yf
import pandas as pd

news_path = "/Users/hibabelhaj/Desktop/LSTMvsTRANSFORMERS/data/raw_news.csv"
news = pd.read_csv(news_path)

tickers = list(news["stock"].value_counts().head(10).index)
print("Tickers:", tickers)

all_data = []

for t in tickers:
    print("Downloading:", t)
    
    # Using Ticker().history() instead of download()
    stock = yf.Ticker(t)
    df = stock.history(
        start="2009-02-14",
        end="2020-06-11",
        interval="1d"
    )

    print("Shape returned:", df.shape)

    if df.empty:
        print("❌ No data for", t)
        continue

    # Reset index to turn Date into column
    df = df.reset_index()

    # Add ticker column
    df["Ticker"] = t

    # Standardize column names
    df = df.rename(columns={
        "Close": "ClosePrice",
        "Open": "OpenPrice",
        "High": "HighPrice",
        "Low": "LowPrice",
        "Volume": "Volume"
    })

    all_data.append(df)

# Combine into long format
prices = pd.concat(all_data, ignore_index=True)

# Drop missing close prices
prices = prices.dropna(subset=["ClosePrice"])

print("\nFinal dataset:")
print(prices.head(10))
print("Final shape:", prices.shape)

save_path = "/Users/hibabelhaj/Desktop/LSTMvsTRANSFORMERS/data/raw_prices.csv"
prices.to_csv(save_path, index=False)

print("\nSaved →", save_path)
```
